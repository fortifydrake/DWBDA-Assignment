### Assignment 3: Take the sales data from anywhere, and write the spark program to calculate no. of product sold in each category

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum

# Initialize SparkSession
spark = SparkSession.builder.appName("SalesDataCategoryCount").getOrCreate()
```

```python
# Load the sales data
df = spark.read.csv("transaction_data.csv", header=True, inferSchema=True)

df.show()
```

```python
# Calculate number of products sold in each category (Country)
category_counts = df.groupBy("Country").agg(sum("NumberOfItemsPurchased").alias("TotalProductsSold")).orderBy("TotalProductsSold", ascending=False)

# Show the results
category_counts.show(truncate=False)
```

```python
# Calculate number of products sold in each category (Items)
category_counts = df.groupBy("ItemDescription").agg(sum("NumberOfItemsPurchased").alias("TotalProductsSold")).orderBy("TotalProductsSold", ascending=False)

# Show the results
category_counts.show(truncate=False)
```

```python
# Stop the Spark session
spark.stop()
```
